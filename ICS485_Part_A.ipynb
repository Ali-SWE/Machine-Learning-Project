## Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from imblearn.metrics import geometric_mean_score
```

## Dataset analysis and report on important statistics

### Reading the dataset

```python
train_data = pd.read_csv('Dataset-train-vf.csv')
test_data = pd.read_csv('Dataset-test-vf.csv')
```

### Statistics

```python
print("Statistical infromation about the training data")
display(train_data.describe())
print("Statistical infromation about the testing data")
display(test_data.describe())
```

```python
print("Number of null values and samples of the training data")
display(train_data.info())
print("Number of null values and samples of the testing data")
display(test_data.info())
```

## Feature selection/transformation/engineering

### Label Encoding for label y

```python
print("for training data")
print("before: unique values for label y: ", train_data['y'].unique())
train_data['y'] = train_data['y'].map({'Low': 0, 'High': 1})
print("after: unique values for label y: ", train_data['y'].unique())

print("for testing data")
print("before: unique values for label y: ", test_data['y'].unique())
test_data['y'] = test_data['y'].map({'Low': 0, 'High': 1})
print("after: unique values for label y: ", test_data['y'].unique())
```

### Label Encoding for feature x14

```python
print("for training data")
print("before: nique values for label x14: ", train_data['x14'].unique())
train_data['x14'] = train_data['x14'].map({'C1': 1, 'C2': 2, 'C3': 3, 'C4': 4})
print("before: nique values for label x14: ", train_data['x14'].unique())

print("for testing data")
print("before: unique values for feature x14: ", test_data['x14'].unique())
test_data['x14'] = test_data['x14'].map({'C1': 1, 'C2': 2, 'C3': 3, 'C4': 4})
print("before: nique values for feature x14: ", test_data['x14'].unique())
```

 ## Dealing with missing values

```python
print("- x2 has 90% of null, so it is dropped")
print("- x6 has small percent of null values, so they will be replaced with the mean\n")
train_data.drop(columns=['x2'], inplace=True)
train_data['x6'] = train_data['x6'].fillna(train_data['x6'].mean())

test_data.drop(columns=['x2'], inplace=True)
test_data['x6'] = test_data['x6'].fillna(train_data['x6'].mean()) # replacing the null values with the mean of the training data
```

## Dealing with imbalanced data

```python
print(train_data['y'].value_counts())
print("Data is balanced, no need to do techinques like oversampling and undersampling")
```

## Data preparation for all classifiers

```python
# Splitting the data into training and validation sets
features = train_data.iloc[:,1:-1] # x1, x2, x3, ..., x14
label = train_data.iloc[:,-1] # y
X_train, X_val, y_train, y_val = train_test_split(features, label, test_size=0.2, random_state=42)
X_test = test_data.iloc[:,1:-1]
y_test = test_data.iloc[:,-1]

# Normalization
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)
```

## Classifiers

### Logistic Regression

#### Training and Validation

```python
# Training
model = LogisticRegression()
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_val)

# Evaluating
accuracy = accuracy_score(y_val, y_pred)
conf_matrix = confusion_matrix(y_val, y_pred)
report = classification_report(y_val, y_pred)
gmean_val = geometric_mean_score(y_val, y_pred, average='macro')

print("Accuracy:", accuracy)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", report)
print("GMean (Validation):", gmean_val)  # Display GMean
```

#### Testing

```python
# Prediction
y_pred = model.predict(X_test)

# Evaluating
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred)
gmean_test = geometric_mean_score(y_test, y_pred, average='macro')

print("Accuracy:", accuracy)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", report)
print("GMean (Test):", gmean_test)  # Display GMean
```

### KNN

#### Training and Validation

```python
# Trying multiple k values and selecting the best one
best_k = None
best_accuracy = 0
best_gmean = 0
best_model = None

for k in range(1, 50):  # Trying k values from 1 to 50
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    accuracy = accuracy_score(y_val, y_pred)
    gmean = geometric_mean_score(y_val, y_pred, average='macro')
    print(f"k={k}, Accuracy={accuracy}, GMean={gmean}")
    if gmean > best_gmean:
        best_k = k
        best_accuracy = accuracy
        best_gmean = gmean
        best_model = model

print("\nBest k:", best_k)
print("Best Accuracy:", best_accuracy)
print("Best GMean:", best_gmean)

# Evaluating the best model
y_pred_best = best_model.predict(X_val)
conf_matrix = confusion_matrix(y_val, y_pred_best)
report = classification_report(y_val, y_pred_best)

print("\nBest k:", best_k)
print("Accuracy:", best_accuracy)
print("GMean:", best_gmean)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", report)
```

#### Testing

```python
y_pred_best = best_model.predict(X_test)
conf_matrix = confusion_matrix(y_test, y_pred_best)
accuracy = accuracy_score(y_test, y_pred_best)
gmean_test = geometric_mean_score(y_test, y_pred_best, average='macro')
report = classification_report(y_test, y_pred_best)

print("----------------------------")
print("\nBest k:", best_k)
print("Accuracy:", accuracy)
print("GMean:", gmean_test)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", report)
```

### Neural Networks

#### Training and Validation

```python
bestNN = [0, '', '', 0, 0]  # Format: [bestGMean, bestAct, bestSol, bestAlpha, bestSize]
NNhiddenSize = [10, 20, 50]
NNActivation = ['identity', 'logistic', 'tanh', 'relu']
NNSolver = ['sgd', 'adam']
NNAlpha = [0.001, 0.01, 0.1]
NNhiddenSize = [10, 20, 50]

for hidden in NNhiddenSize:
    for acti in NNActivation:
        for sol in NNSolver:
            for alpha in NNAlpha:
                model = MLPClassifier(
                    hidden_layer_sizes=(hidden,),
                    activation=acti,
                    solver=sol,
                    alpha=alpha,
                    max_iter=2000,
                    early_stopping=True,
                    random_state=42
                )
                model.fit(X_train, y_train)
                y_val_pred = model.predict(X_val)
                modelGMean = geometric_mean_score(y_val, y_val_pred, average='macro')

                if modelGMean > bestNN[0]:
                    bestNN = [modelGMean, acti, sol, alpha, hidden]

print('Best Neural Network parameters:', bestNN)

# Train final model with the best parameters
best_model = MLPClassifier(
    hidden_layer_sizes=(bestNN[4],),
    activation=bestNN[1],
    solver=bestNN[2],
    alpha=bestNN[3],
    max_iter=2000,
    early_stopping=True,
    random_state=42
)
best_model.fit(X_train, y_train)

# Evaluate on validation set
y_val_pred = best_model.predict(X_val)
val_accuracy = accuracy_score(y_val, y_val_pred)
val_gmean = geometric_mean_score(y_val, y_val_pred, average='macro')
val_conf_matrix = confusion_matrix(y_val, y_val_pred)

print("Validation Accuracy:", val_accuracy)
print("Validation GMean:", val_gmean)
print("Validation Confusion Matrix:\n", val_conf_matrix)
```

#### Testing

```python
# Evaluate on test set
y_test_pred = best_model.predict(X_test)
test_accuracy = accuracy_score(y_test, y_test_pred)
test_gmean = geometric_mean_score(y_test, y_test_pred, average='macro')
test_conf_matrix = confusion_matrix(y_test, y_test_pred)

print("Test Accuracy:", test_accuracy)
print("Test GMean:", test_gmean)
print("Test Confusion Matrix:\n", test_conf_matrix)
```

### Adaboost

#### Training and Validation

```python
# Hyperparameter grid for AdaBoost
RFNestimators = [2, 5, 10, 20, 50]
RFlearning = [0.01, 0.1, 1, 10]

# Store best hyperparameters and GMean
bestRF = [0, 0, 0]

# Test different hyperparameters
for nest in RFNestimators:
    for learn in RFlearning:
        model = AdaBoostClassifier(
            estimator=DecisionTreeClassifier(criterion='gini', max_depth=3),
            n_estimators=nest,
            learning_rate=learn
        )
        model.fit(X_train, y_train)
        y_val_pred = model.predict(X_val)
        gmean = geometric_mean_score(y_val, y_val_pred, average='macro')

        if gmean > bestRF[0]:
            bestRF = [gmean, nest, learn]

print("Best hyperparameters:", bestRF)

# Train final model with best parameters on the training data
AdaModel = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(criterion='gini', max_depth=3),
    n_estimators=bestRF[1],
    learning_rate=bestRF[2]
)
AdaModel.fit(X_train, y_train)

# Evaluate on validation set
y_val_pred = AdaModel.predict(X_val)
val_gmean = geometric_mean_score(y_val, y_val_pred, average='macro')
val_accuracy = accuracy_score(y_val, y_val_pred)
val_conf_matrix = confusion_matrix(y_val, y_val_pred)

print("Validation GMean:", val_gmean)
print("Validation Accuracy:", val_accuracy)
print("Validation Confusion Matrix:\n", val_conf_matrix)
```

#### Testing

```python
# Evaluate on test set
y_test_pred = AdaModel.predict(X_test)
test_gmean = geometric_mean_score(y_test, y_test_pred, average='macro')
test_accuracy = accuracy_score(y_test, y_test_pred)
test_conf_matrix = confusion_matrix(y_test, y_test_pred)

print("----------------------------")
print("Test GMean:", test_gmean)
print("Test Accuracy:", test_accuracy)
print("Test Confusion Matrix:\n", test_conf_matrix)
```
